```python
import requests
headers={'User-Agent':'Mozilla/5.0(Windows NT 10.0;Win64;x64)AppleWebKit/537.36(KHTML,like Gecko)Chrome/69.0.3497.100 Safari/537.36'}
url='https://news.ecust.edu.cn/18/list.htm'
res=requests.get(url,headers=headers)
res.encoding="utf-8"
res=res.text

print(res)
```

```python
import re
p_time1='<span class="times">(.*?)</span>'
time1=re.findall(p_time1,res,re.S)
print(time1)
```

```python
import re
p_href='<div class="news_title"><a href="(.*?)"'
href=re.findall(p_href,res,re.S)
print(href)
```

```python
import re
p_title='<div class="news_title"><a href=".*?" target="{打开方式}" title=".*?">(.*?)</a></div>'
title=re.findall(p_title,res,re.S)
print(title)
```

```python
for i in range(len(title)):
    title[i]=title[i].replace('\u200b','')
    print(title[i])
```

```python
for i in range(len(title)):
    print('['+str(i+1)+']'+title[i])
    print('https://news.ecust.edu.cn/18/list'+href[i])
    print(time1[i])
```

```python
file1=open('数据挖掘报告.txt','a')
for i in range(len(title)):
    file1.write('['+str(i+1)+']'+title[i]+'\n')
    file1.write('https://news.ecust.edu.cn/18/list'+href[i]+'\n')
    file1.write('\t'+time1[i]+'\n')
file1.close()
```

```python
import time

while True:
    file1=open('数据挖掘报告.txt','a')
    for i in range(len(title)):
        file1.write('['+str(i+1)+']'+title[i]+'\n')
        file1.write('https://news.ecust.edu.cn/18/list'+href[i]+'\n')
        file1.write('\t'+time1[i]+'\n')
    file1.close()
    time.sleep(10800)
```

```python
import requests
from time import sleep
headers={'User-Agent':'Mozilla/5.0(Windows NT 10.0;Win64;x64)AppleWebKit/537.36(KHTML,like Gecko)Chrome/69.0.3497.100 Safari/537.36'}

start_page=1
end_page=10#假设爬取到第十页
for page_num in range(start_page,end_page+1):
    url=f'https://news.ecust.edu.cn/18/list{page_num}.htm'

    try:
        res=requests.get(url,headers=headers)
        res.encoding="utf-8"
        res=res.text
        print(f"正在爬取第{page_num}页的内容")
        print(res)
    except Exception as e:#异常处理
        print(f"爬取第{page_num}页时出错：{e}")
    sleep(1)#每次请求后延时一秒避免请求过于频繁
```
